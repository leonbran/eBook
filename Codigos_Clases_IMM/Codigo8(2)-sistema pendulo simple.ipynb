##Sistema de competicion tipo Lotka-Volterra


El modelo de competición de Lotka-Volterra, también conocido como el modelo de competencia de dos especies, es una extensión del modelo clásico de presa-depredador de Lotka-Volterra que describe la interacción entre dos especies que compiten por un recurso común en un ecosistema.

Supongamos que tenemos dos especies de animales, conejos (especie 1) y ovejas (especie 2), que compiten por la misma fuente de alimento. Podemos describir la dinámica de estas poblaciones utilizando el modelo de competición de Lotka-Volterra.

Supongamos que las ecuaciones para este modelo podrían ser:

<br>

\begin{eqnarray}
\frac{dx}{dt} &=& x(3 - x - 2y),\\
\frac{dy}{dt} &=& y(2 -  x - y).
\end{eqnarray}

<br>

donde:

<br>

- $x$ es la población de conejos (especie 1),
- $y$ es la población de zorros (especie 2),
- Los parámetros $3$, $-1$, $-2$ y $2$, $-1$ y $-1$ son valores arbitrarios que representan tasas de crecimiento intrínseco y coeficientes de competencia.


<br>


Este modelo puede simular cómo las poblaciones de conejos y las ovejas interactúan a lo largo del tiempo, reflejando el impacto de la competencia y la depredación en sus números respectivos. Dependiendo de los valores específicos de los parámetros, podemos observar diferentes comportamientos, como la coexistencia estable de ambas especies, la exclusión competitiva de una especie por la otra, o ciclos de población en las que las poblaciones fluctúan periódicamente.

<br>

Resolviendo las ecuaciones se puede probar que los puntos criticos son: (0,0), (0,2), (3,0) y (1,1). Con la informacion anterior, realize el  campo de velocidades y grafique los puntos criticos

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir las ecuaciones del sistema
def lotka_volterra_competicion(x,y):

    # Definir parámetros del sistema

    r1=3
    r2=2
    a11=-1
    a12=-2
    a21=-1
    a22=-1

    dx_dt = x * (r1 + a11 * x + a12 * y)
    dy_dt = y * (r2 + a21 * x + a22 * y)

    return  dx_dt, dy_dt


# Definir la malla de puntos para evaluar el campo de velocidades
x = np.linspace(-3.5,5, 20)
y = np.linspace(-2.5, 5, 20)
X, Y = np.meshgrid(x, y)




# Calcular las velocidades en cada punto de la malla
dx_dt, dy_dt = lotka_volterra_competicion( X, Y )

norm = np.sqrt( dx_dt**2 + dy_dt**2)
dx_dt  = dx_dt / norm
dy_dt  = dy_dt / norm



# Graficar el campo de velocidades
plt.figure(figsize=(8, 6))
plt.quiver(X, Y, dx_dt, dy_dt )
#plt.streamplot(X, Y, dx_dt, dy_dt , density=1.5 )

plt.xlabel('Población de conejos (x)')
plt.ylabel('Población de ovejas (y)')
plt.title('Campo de velocidades del modelo de competición de Lotka-Volterra')



# El punto (gamma/delta,alpha/beta)
plt.plot( 0 , 0, 'ro' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 0 , 2, 'bo' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 3 , 0, 'mo' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 1 , 1, 'go' , markersize=15 )



plt.grid()
plt.show()

```

**Opcional**
Tambien podemos usar el streamplot!

```python

# Graficar el campo de velocidades
plt.figure(figsize=(8, 6))
plt.streamplot(X, Y, dx_dt, dy_dt , density=1.5 )

plt.xlabel('Población de conejos (x)')
plt.ylabel('Población de ovejas (y)')
plt.title('Campo de velocidades del modelo de competición de Lotka-Volterra')



# El punto (gamma/delta,alpha/beta)
plt.plot( 0 , 0, 'ro' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 0 , 2, 'bo' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 3 , 0, 'mo' , markersize=15 )


# El punto (gamma/delta,alpha/beta)
plt.plot( 1 , 1, 'go' , markersize=15 )



plt.grid()
plt.show()
```

## Campo de velocidades de una ecuación de segundo orden

De acuerdo con la fisica newtoniana, el movimiento del sistema modelo llamado [péndulo simple](https://es.wikipedia.org/wiki/P%C3%A9ndulo_simple) se puede describir a través de la ecuación diferencial

\begin{equation}
\dfrac{d^2 \theta}{d t^2} + \dfrac{g}{l}\,\sin\theta  = 0,
\end{equation}

donde $\theta$ (*theta*) representa el ángulo con la vertical, $g$ la acelaración de la gravedad y $l$ la longitud del péndulo.



Esta ecuación es difícil de solucionar de forma analítica pues se trata de una [ecuación diferencial no lineal](https://es.wikipedia.org/wiki/Ecuaci%C3%B3n_diferencial#Ecuaciones_diferenciales_no_lineales).  Sin embargo, esta  ecuación diferencial de segundo orden se puede representar por el sistema de ecuaciones de primer orden:

\begin{eqnarray}
\dfrac{d \theta}{d t} &=& \omega, \\
\dfrac{d \omega}{d t} &=& -\dfrac{g}{l}\sin\theta,
\end{eqnarray}

en el que se ha definido una nueva variable $\omega = \dfrac{d\theta}{d t}$ (velocidad angular, *omega*).

El campo de velocidades para este sistema en el plano $(\theta, \omega)$ en cierto punto $(\theta_0, \omega_0)$ se puede definir como  

\begin{equation}
\vec{u} = (u_1 , u_2) :=\dfrac{1}{\sqrt{ V_1^2 +  V_2^2 } } ( V_1 ,  V_2),
\end{equation}

donde
- $V_1 = \dfrac{d \theta}{dt}\Big|_{\theta_0, \omega_0} = \omega_0$

y  

- $V_2 = \dfrac{d \omega}{dt} \Big|_{\theta_0, \omega_0} = -\frac{g}{l}\,\sin\theta_0$.

### Problema:

Suponiendo que $l = 1~\text{m}$ y $g = 9.8~\text{m/s}^2$,  grafique el campo de velocidades de este sistema para $\theta \in [-\pi, \pi]$ y  $\omega \in [-2 \pi, 2 \pi]$.



### Solución



```python
import numpy as np
import matplotlib.pyplot as plt

# Sistema de ecuaciones diferenciales
# para las variables
# ángulo (theta)
# velocidad angular (omega)

L = 1.0
g = 9.8
def edo_pendulo(angulo,velocidad):
  dtheta_dt = velocidad
  domega_dt = -(g/L)*np.sin(angulo)
  return dtheta_dt, domega_dt
```

```python
# Definición de parámetros y
# construcción del mallado
theta_min  = -np.pi    # theta mínimo
theta_max  =  np.pi    # theta máximo
omega_min  = -2*np.pi  # vel. angular mínima
omega_max  =  2*np.pi  # vel. angular máxima

N = 40
theta  = np.linspace(theta_min,theta_max, N)  # Valores iniciales variable theta
omega  = np.linspace(omega_min,omega_max, N)  # Valores iniciales variable omega


#El paso mas importante!
theta_malla, omega_malla = np.meshgrid(theta, omega)    # Crea el mallado (theta, omega)
```

```python
# Evaluación del campo de velocidades
V1, V2 = edo_pendulo(theta_malla, omega_malla) # Componentes vector velocidad (V)
norma = np.sqrt( V1**2 + V2**2 )        # Norma vector
### Vector unitario (u)
u1 = V1 / norma
u2 = V2 / norma

# Grafica el campo de velocidades
#plt.streamplot(theta_malla, omega_malla, u1, u2, density = 1.6,color=norma)
#plt.colorbar(label='Norma')
plt.streamplot(theta_malla, omega_malla, u1, u2, density = 1.6)
plt.title('Campo de velocidades péndulo')
plt.xlabel(r'Ángulo ($\theta$)')
plt.ylabel(r'Velocidad angular ($\omega$)')
#plt.axis([theta_min, theta_max, omega_min, omega_max])
plt.grid()
plt.show()
```
