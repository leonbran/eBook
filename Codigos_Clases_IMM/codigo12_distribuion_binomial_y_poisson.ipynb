##Distribucion binomial
---

Podemos usar la libreria scipy para ello!

- Graficar la función de masa de probabilidad de la distribución binomial

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Definir los parámetros de la distribución binomial
n = 10  # Número total de ensayos
p = 0.5 # Probabilidad de éxito en cada ensayo


# Calcular la función de masa de probabilidad (PMF) teórica
x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

# Graficar el histograma de la muestra y la PMF teórica
plt.bar(x, pmf,  alpha=0.5)



plt.xlabel('Número de Éxitos')
plt.ylabel('Probabilidad')
plt.title('Distribución Binomial')

plt.show()
```

## Distribución de Poisson
----

####Ejemplo

Supongamos que se realizan $10$ experimentos en un laboratorio para estudiar la tasa de reacciones químicas en una solución. Se registró el número  de reacciones quimicas en cada uno de los experimentos durante 1 minuto y se obtuvieron las siguientes reacciones en cada solución:

<br>

|   Experimentos   | Éxitos |
|:-----------:|:------:|
|  Experimento 1  |   20  |
|  Experimento 2  |   40   |
|  Experimento 3  |   30   |
|  Experimento 4  |   60   |
|  Experimento 5  |   50   |
|  Experimento 6  |   20   |
|  Experimento 7  |   10   |
|  Experimento 8  |   30  |
|  Experimento 9  |   40   |
| Experimento  10 |   20   |


<br>

A partir de los datos obtenidos se encuentra que la tasa media de reacciones por minuto, denotada como $\lambda$, es de $32.0$ reacciones/min. Con la informacion anterior:

<br>

1. Encuentre la probabilidad porcentual de que ocurran $1920$ reaciónes quimicas en una hora.

<br>

2. Realice un gráfico en donde se muestre cual es el numero de reacciones cada 2 minutos en un intervalo de reacciones entre $10$ y $100$.


###Solución


<br>

1.  Aqui debemos cambiar la escala del parametro $\lambda$ de  $\dfrac{reacciones}{ minuto}$ a  $\dfrac{reacciones}{ hora}$.

Para ello, simplemente debemos multiplicar a lambda_estimado por $60$. Despues de esto, usamos la distribucion de Poisson para estimar la probabilidad requerida.


$$\lambda \dfrac{[reacciones]}{[hora]} =  \lambda \dfrac{[reacciones]}{[minuto]} \times \dfrac{[60 minutos]}{[hora]}$$

<br>


```python
import numpy as np
from scipy.stats import poisson


# Cálculo de la probabilidad para obtener 1920 reacciones quimicas
valor_variable_aleatoria = 1920



#Reacciones quimicas por minuto
lambda_estimado = 32


#Reacciones quimicas por minuto
nuevo_lambda = 32 * 60

probabilidad_2 = poisson.pmf(valor_variable_aleatoria, nuevo_lambda)




print("probabilidad porcentual para obtener 1920 reacciones quimicas = ",  100*probabilidad_2 , "%")
```

<br>

- Gráfico  

2. Aqui debemos multiplicar la constante Lambda por 2, para obtener numero de reaccions por 2 minutos!

$$\lambda \dfrac{[reacciones]}{[2 minutos]} =  2  \lambda \dfrac{[reacciones]}{[minuto]}  $$


```python
#Aqui debemos multiplicar la constante Lambda por 2, para obtener numero de reaccions por 2 minutos!

import matplotlib.pyplot as plt

nuevo_lambda = lambda_estimado * 2

for posibles_valores in range( 10, 100):

  #Usaremos barras para graficar probabilidad porcentual
  plt.bar( posibles_valores ,    poisson.pmf( posibles_valores , nuevo_lambda) , color='green' )


plt.ylabel("p(y)")
plt.xlabel("Numero de reacciones quimicas por cada dos minutos")



#valor esperado
mu = nuevo_lambda


#Note que debemos tomar la parte entera de mu!
plt.scatter( nuevo_lambda , poisson.pmf( nuevo_lambda , nuevo_lambda) , color='r' )
plt.grid()
plt.show()

```
