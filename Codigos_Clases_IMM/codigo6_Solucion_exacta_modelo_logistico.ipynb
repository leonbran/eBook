#Modelo exponencial

###Modelo exponencial discreto

El modelo exponencial discreto es (sucesion recurrente):

$$ P_{n+1} = \kappa P_{n}$$


Y la respectiva ecuacion en diferencias es:

$$\Delta P_{n } = (\kappa-1) P_{n}$$


```python

#Generemos algunos puntos para el modelo exponencial logistico

import numpy as np

def Pn(n):

    #definimos el valor inicial
    Pn=[1]

    kappa = 1.1

    for i in range(n):

        Pn.append( Pn[i]*( kappa ) )

    return Pn

Pn_puntos = Pn(30)
n_puntos  = [i for i in range( len(Pn_puntos ) )]


#graficando
plt.plot( n_puntos , Pn_puntos , 'bo' , label='Datos de la ecución en diferencias' )

```

###Modelo exponencial continuo

Ecuacion diferencial modelo exponencial es (¿por que? venga a clase!):

$$ \dfrac{dP(t)}{dt} =  (  \kappa -1  ) P(t)$$

Lo cual resulta por el metodo de separación de variables en (toca venir a clase!)

<br>

$$ \boxed{ P(t)= P_{0} e^{(k-1)t} } $$

<br>

Con $P_{0}$ una constante de integración.

### Comparacion de modelo continuo y discreto

```python
import matplotlib.pyplot as plt
import numpy as np


kappa = 1.1
Po   = 1


t = np.linspace(0,30,100)

#Solucion Exacta!
P = Po * np.exp( (kappa-1) * t  )



plt.plot( t , P  , 'g-' , label='solucion EDO' )



#graficando
plt.plot( n_puntos , Pn_puntos , 'bo' , label='Datos de la ecución en diferencias' )


plt.xlim([0,15])
plt.ylim([1,5])


plt.xlabel('t')
plt.ylabel('P')
plt.legend()
plt.grid()

plt.show()
```

#La ecuación logistica

###Modelo logistico discreto

El modelo logistico discreto se define por la siguiente ecuacion de recurrencia

$$P_{n+1} = \kappa P_{n} -  \beta P_{n}^2$$

Y la respectiva ecuacion en diferencias es:

$$ \Delta P_{n+1} = (\kappa-1) P_{n} -  \beta P_{n}^2 $$




```python
import numpy as np

def Pn(n):

    #definimos el valor inicial
    Pn=[1]

    kappa = 1.5

    beta =  0.001

    for i in range(n):

        Pn.append( Pn[i]*( kappa ) - beta *  Pn[i]**2 )

    return Pn

Pn_puntos = Pn(30)
n_puntos  = [i for i in range( len(Pn_puntos ) )]


plt.plot( n_puntos , Pn_puntos , 'bo' , label='datos_originales' )
plt.grid()
plt.show()
```

###Modelo logistico continuo

La ecuacion diferencial del modelo logistico continuo es (¿por que?):

$$ \dfrac{dP(t)}{dt} = P(t) ( \kappa-1 )  - \beta P(t)^2    $$



Lo cual resulta por el metodo de separacion de variables en (ver notas de clase!)

<br>

$$ \boxed{  P(t)=  \frac{(k-1) e^{kt}}{(-\beta+k-1 ) e^{  t} + \beta e^{tk}} } $$

<br>

Con $c$ una constante de integración.

###Comparacion de modelo continuo y discreto

Por tanto escogiendo las constantes:  $\kappa=1.5$ y $\beta=0.001$ obtenemos el grafico  

```python
import matplotlib.pyplot as plt
import numpy as np


kappa = 1.5
beta  = 0.001

t = np.linspace(0,30,100)

#Solucion Exacta!
P =  (  (kappa-1) * np.exp( (kappa ) * t  ) ) / (  np.exp(t) * (-beta + kappa -1 ) + beta * np.exp(kappa * t)  )

plt.plot( t , P  , 'g-' , label='solucion EDO' )



#graficando
plt.plot( n_puntos , Pn_puntos , 'bo' , label='datos_originales' )


plt.xlabel('t')
plt.ylabel('P')
plt.legend()
plt.grid()

plt.show()
```

Comentario importante: Note que el modelo continuo y el modelo discreto son diferentes en los lugares    las derivadas de la funcion son grandes. Esto se debe principalemnte porque la aproximación

$$ \Delta P_n \not\approx \dfrac{dP(t)}{dt} $$
