#Campo de velocidades de un sitema de ODEs



La evaluación del [campo de velocidades](http://faculty.sfasu.edu/judsontw/ode/html-20190821/systems02.html) permite obtener una idea geométrica general del comportamiento de un sistema cuya dinámica se modela a través de una ecuación diferencial dada. Suponga que se quiere analizar un sistema de EDO de la forma

\begin{eqnarray}
\dfrac{dx(t)}{dt} &=& f(x,y)\\
\dfrac{dy(t)}{dt} &=& g(x,y)
\end{eqnarray}

El campo de velocidades en el plano $(x,y)$ se define asignando a cada punto $(x_0,y_0)$  el vector unitario
\begin{equation}
\vec{u} = (u_1 , u_2) := \dfrac{1}{\sqrt{ V_1^2 + V_2^2 } } \, ( V_1 , V_2) ,
\end{equation}
donde
- $V_1 = \dfrac{dx}{dt} \Big|_{x_0,y_0} = f(x_0,y_0)$

- $V_2 = \dfrac{dy}{dt} \Big|_{x_0,y_0} = g(x_0,y_0)$

$~$

El campo de velocidades nos sirve para obtener información "cualitativa" del comportamiento de la solución. En particular podemos obtener información de:

- puntos de equilibrio.
- puntos de silla.
- regiones del plano de crecimiento y decrecimiento dependiendo de la condición inicial $(x_0,y_0)$

**Nota**: Es importante notar que no es necesario resolver la ecuación diferencial para hallar su campo de velocidades

<br>


##Ejemplo: sistema de Lotka-Volterra

El modelo depredador-presa, conocido como el modelo de Lotka-Volterra, fue desarrollado de manera independiente por Alfred J. Lotka y Vito Volterra en la década de 1920. Estos modelos   se han convertido en una base fundamental para el estudio de las interacciones entre especies en ecología y la comprensión de las dinámicas poblacionales.  Su expresion matematica es la siguiente:


<br>

\begin{align*}
\text{Para las presas (P):} \quad \frac{{dP}}{{dt}} &= P( \alpha  - \beta   D) \\
\text{Para los depredadores (D):} \quad \frac{{dD}}{{dt}} &= - D( \gamma - \delta P ) \
\end{align*}

<br>

Donde:

<br>

$\frac{{dP}}{{dt}}$  y $\frac{{dD}}{{dt}}$ representan las tasas de cambio de las poblaciones de presas y depredadores, respectivamente, con respecto al tiempo.

<br>

- P   y   D   son las poblaciones de presas y depredadores, respectivamente.

- $\alpha$   es la tasa de crecimiento de las presas en ausencia de depredadores.   
- $\beta$   es la tasa de mortalidad de las presas debido a la depredación.

- $\delta$ es la tasa de crecimiento de los depredadores en función de las
presas.  

- $\gamma$ es la tasa de mortalidad de los depredadores.


<br>
<br>

**Ejercicio**

Hacer el campo de velocidades y graficar los puntos criticos

```python
import numpy as np
import matplotlib.pyplot as plt


#-------------parametros del sistema------------------
alpha = 1.2  # Tasa de crecimiento de las presas
beta = 0.6   # Tasa de mortalidad de las presas debido a la depredación

gamma = 0.3  # Tasa de mortalidad de los depredadores
delta = 0.8  # Tasa de crecimiento de los depredadores en función de las presas
#-----------------------------------------------------


# Definir el sistema de ecuaciones del modelo Lotka-Volterra
def lotka_volterra( P , D ):

    dP_dt = alpha * P     - beta * P * D

    dD_dt = delta * P * D - gamma * D

    return  dP_dt, dD_dt


# Crear una cuadrícula de puntos en el espacio de fase
P_range = np.linspace(0,   1 , 10)
D_range = np.linspace(0,   4 , 10)
P, D = np.meshgrid(P_range, D_range)


#Que es P y D?

```

```python
#veamos que es P
print( np.round(P,2))
```

```python
#veamos que es D
print( np.round(D,2) )
```

```python
# Calcular las velocidades en cada punto de la cuadrícula
dP_dt, dD_dt = lotka_volterra( P, D )


print( "dP_dt = ", np.round(dP_dt,3)  )
print(" ")
print( "dD_dt = ", np.round(dD_dt,3) )
```

```python


# Normalizar las velocidades para que los vectores tengan la misma longitud
norm = np.sqrt( dP_dt**2 + dD_dt**2)
dP_dt  = dP_dt /norm
dD_dt  = dD_dt /norm


# Trazar el campo de velocidades
plt.figure(figsize=(8, 6))
plt.quiver(P, D, dP_dt, dD_dt  )
plt.xlabel('Presas (P)')
plt.ylabel('Depredadores (D)')
plt.title('Campo de Velocidades - Modelo Lotka-Volterra')
plt.grid(True)



# Trazemos los puntos criticos

# El punto (0,0)
plt.plot( 0, 0, 'ko' )



# El punto (gamma/delta,alpha/beta)
plt.plot( gamma/delta , alpha/beta, 'ro' , markersize=15 )

# Mostrar todos los graficos juntos
plt.show()
```

NOTA: Los puntos crítico de extinción;  Este punto crítico ocurre cuando la población de presas o depredadores es cero, lo que indica la extinción de una de las especies en el sistema. Se representan como (0, D) , (P, 0) , (0, 0).
