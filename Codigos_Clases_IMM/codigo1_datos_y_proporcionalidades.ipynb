## Datos inversamente proporcionales

Se asume la relación: $$y = \dfrac{C}{x}$$

```python
import numpy as np

import matplotlib.pyplot as plt



# Datos de la tabla
x = np.linspace(1,10,10)

C= 3

y = C/x



# Crear la gráfica
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Relación inversamente proporcional')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Relación Inversamente Proporcional')
plt.legend()
plt.grid(True)

# Mostrar la gráfica
plt.show()
```
