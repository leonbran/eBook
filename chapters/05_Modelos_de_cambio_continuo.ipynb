# Modelos de cambio continuo

En el capítulo anterior estudiamos modelos de cambio discreto, en los cuales el estado de un sistema se describe mediante una sucesión

$$
x_0,x_1,x_2,\ldots
$$

y su evolución mediante una regla como

$$
x_{n+1}=f(x_n).
$$

Esta descripción es adecuada cuando observamos un fenómeno en instantes separados de tiempo. Sin embargo, muchos fenómenos evolucionan de manera continua, como el crecimiento de una población, la desintegración de una sustancia o el movimiento de una partícula.

En estos casos, el estado del sistema se representa mediante una función

$$
x=x(t),
$$

y su evolución mediante una relación entre la función y su tasa de cambio:

$$
\frac{dx}{dt}=f(x,t).
$$

Estas relaciones se denominan ecuaciones diferenciales ordinarias (EDO), y las discutiremos mas adelante. En este capítulo construiremos y analizaremos modelos continuos, estudiaremos sus puntos críticos y campos de velocidades, y utilizaremos métodos numéricos para obtener y visualizar soluciones aproximadas.



## 1. De modelos discretos a continuos



### 1.1. Cambio discreto

En un modelo discreto estudiamos el cambio entre dos instantes consecutivos mediante

$$
x_{n+1}-x_n.
$$

Si los instantes están separados por un intervalo de duración $\Delta t$, podemos considerar la tasa de cambio promedio

$$
\frac{x_{n+1}-x_n}{\Delta t}.
$$

Esta expresión indica cuánto cambia, en promedio, la cantidad $x$ por unidad de tiempo durante el intervalo. Por ejemplo, si una población pasa de $1000$ a $1200$ individuos durante un año, entonces

$$
\frac{x_{n+1}-x_n}{\Delta t}
=
\frac{1200-1000}{1}
=
200.
$$

La población aumentó, en promedio, en $200$ individuos por año durante ese intervalo.



### 1.2. Tasa de cambio instantánea

La tasa de cambio promedio depende del tamaño del intervalo considerado. Si queremos conocer la tasa de cambio en un instante particular, debemos reducir progresivamente el intervalo. Para una función continua $x(t)$, la derivada se define mediante

$$
\frac{dx}{dt}
=
\lim_{\Delta t\to0}
\frac{x(t+\Delta t)-x(t)}{\Delta t}.
$$

La derivada representa entonces la **tasa de cambio instantánea** de la variable $x$ respecto del tiempo. Esta interpretación es fundamental para el modelamiento.

Por ejemplo:

- $\dfrac{dN}{dt}$ representa la tasa de cambio de una población;
- $\dfrac{dC}{dt}$ representa la tasa de cambio de una concentración;
 $\dfrac{dv}{dt}$ representa la aceleración de una partícula.

La derivada permite pasar de una descripción basada en cambios entre dos mediciones a una descripción del comportamiento instantáneo del sistema.

Esta diferencia entre cambio discreto y cambio continuo es fundamental para distinguir los dos tipos de modelos: mientras los modelos discretos describen la evolución mediante cambios entre estados sucesivos, los modelos continuos describen la evolución mediante tasas de cambio instantáneas.


### 1.3. De una ecuación discreta a una ecuación diferencial

Consideremos el modelo discreto

$$
x_{n+1}=x_n+r x_n\Delta t.
$$

Restando $x_n$,

$$
x_{n+1}-x_n=r x_n\Delta t.
$$

Dividiendo entre $\Delta t$,

$$
\frac{x_{n+1}-x_n}{\Delta t}=r x_n.
$$

Si hacemos que $\Delta t$ sea cada vez más pequeño, la diferencia cociente se aproxima a una derivada:

$$
\frac{dx}{dt}=rx.
$$

Hemos obtenido una ecuación diferencial a partir de un modelo de cambio discreto.

Este procedimiento ilustra una conexión fundamental:

$$
\boxed{
\frac{x_{n+1}-x_n}{\Delta t}
\longrightarrow
\frac{dx}{dt}
}
$$

cuando

$$
\Delta t\rightarrow0.
$$



* #### Ejemplo: crecimiento proporcional

Supongamos que una población $N(t)$ aumenta a una tasa proporcional a su tamaño.

En un modelo discreto habíamos utilizado

$$
N_{n+1}-N_n=rN_n.
$$

En el modelo continuo, la hipótesis de crecimiento proporcional se expresa mediante

$$
\frac{dN}{dt}=rN.
$$

La interpretación es directa:

> cuanto mayor es la población, mayor es su tasa de crecimiento.

Si la población es el doble, la tasa de crecimiento también es el doble.

Esta ecuación será uno de los modelos fundamentales de este capítulo.




## 2. Fundamentos matemáticos: ecuaciones diferenciales ordinarias



### 2.1. ¿Qué es una ecuación diferencial?

Una ecuación diferencial es una ecuación que relaciona una función desconocida con una o más de sus derivadas.

Por ejemplo,

$$
\frac{dx}{dt}=3x
$$

es una ecuación diferencial ordinaria porque relaciona la función $x(t)$ con su primera derivada. Otro ejemplo es

$$
\frac{d^2x}{dt^2}+x=0.
$$

En este caso aparece la segunda derivada. Una ecuación diferencial ordinaria contiene derivadas respecto de una sola variable independiente. Por ello se denomina **ordinaria**.

En este capítulo nos concentraremos principalmente en ecuaciones de primer orden de la forma

$$
\frac{dx}{dt}=f(t,x).
$$



### 2.2. Orden de una ecuación diferencial

El **orden** de una ecuación diferencial corresponde al orden de la derivada de mayor grado que aparece en ella.

Por ejemplo,

$$
\frac{dx}{dt}=x
$$

es una ecuación de primer orden, mientras que

$$
\frac{d^2x}{dt^2}+4\frac{dx}{dt}+3x=0
$$

es una ecuación de segundo orden.

Los modelos de cambio continuo que estudiaremos inicialmente serán principalmente de primer orden.



### 2.3. Solución de una ecuación diferencial

Una solución de una ecuación diferencial es una función que satisface la ecuación en un intervalo determinado.

Por ejemplo, consideremos

$$
\frac{dx}{dt}=x.
$$

La función

$$
x(t)=e^t
$$

es una solución porque

$$
\frac{dx}{dt}=e^t=x(t).
$$

Pero también lo es

$$
x(t)=2e^t,
$$

pues

$$
\frac{dx}{dt}=2e^t=x(t).
$$

En realidad, la familia

$$
x(t)=Ce^t
$$

es una familia de soluciones, donde $C$ es una constante.

Esto plantea una pregunta importante: ¿cómo seleccionamos una solución particular?

La respuesta nos lleva a los problemas de valor inicial.




## 3. Problemas de valor inicial




Un modelo continuo normalmente contiene dos componentes:

1. una ecuación que describe la regla de evolución;
2. una condición inicial que especifica el estado del sistema en un instante determinado.

Por ejemplo,

$$
\frac{dx}{dt}=f(t,x),
$$

junto con

$$
x(t_0)=x_0.
$$

Este problema se denomina **problema de valor inicial**.

Podemos representarlo como

$$
\boxed{
\text{ecuación diferencial}
+
\text{condición inicial}
=
\text{problema de valor inicial}
}
$$

La ecuación describe cómo evoluciona el sistema, mientras que la condición inicial determina desde qué estado comienza.



* #### Ejemplo de aplicación

Consideremos

$$
\frac{dx}{dt}=x,
$$

con

$$
x(0)=2.
$$

Sabemos que la solución general es

$$
x(t)=Ce^t.
$$

Utilizando la condición inicial,

$$
x(0)=Ce^0=C=2.
$$

Por tanto,

$$
\boxed{x(t)=2e^t}.
$$

La condición inicial ha permitido seleccionar una única trayectoria dentro de la familia de soluciones.



## 4. Método de separación de variables



### 4.1. Ecuaciones separables

Una ecuación diferencial de primer orden

$$
\frac{dx}{dt}=f(t,x)
$$

se denomina **separable** si puede escribirse en la forma

$$
\frac{dx}{dt}=g(t)h(x).
$$

En ese caso podemos reorganizarla como

$$
\frac{1}{h(x)}\,dx=g(t)\,dt.
$$

De esta manera, las variables $x$ y $t$ quedan separadas.



### 4.2. Procedimiento

Para resolver una ecuación separable:

1. escribir la ecuación en forma separable;
2. separar las variables;
3. integrar ambos lados;
4. utilizar la condición inicial, si existe;
5. despejar la función cuando sea posible.



* #### Ejemplo: Crecimiento exponencial

Consideremos

$$
\frac{dN}{dt}=rN.
$$

Suponiendo $N>0$, podemos escribir

$$
\frac{1}{N}\,dN=r\,dt.
$$

Integramos:

$$
\int\frac{1}{N}\,dN
=
\int r\,dt.
$$

Obtenemos

$$
\ln N=rt+C.
$$

Exponentiando,

$$
N=Ce^{rt}.
$$

Por tanto, la solución general es

$$
\boxed{N(t)=Ce^{rt}}.
$$

Si además

$$
N(0)=N_0,
$$

entonces

$$
C=N_0,
$$

y obtenemos

$$
\boxed{N(t)=N_0e^{rt}}.
$$

Este resultado será la base del modelo de crecimiento exponencial continuo.



* #### Ejemplo: Decaimiento de sustancias

Supongamos que una sustancia pierde una cantidad proporcional a la cantidad presente:

$$
\frac{dC}{dt}=-kC,
$$

donde $k>0$.

Separando variables,

$$
\frac{1}{C}\,dC=-k\,dt.
$$

Integrando,

$$
\ln C=-kt+C_1.
$$

Por tanto,

$$
C(t)=C_0e^{-kt}.
$$

Si inicialmente

$$
C(0)=100,
$$

y

$$
k=0.2,
$$

entonces

$$
\boxed{C(t)=100e^{-0.2t}}.
$$

La cantidad disminuye continuamente, pero nunca se hace exactamente igual a cero.





## 5. Analisis cualitativo de modelos continuos

En muchos problemas de interés no es sencillo obtener una solución analítica de la ecuación diferencial. Además, cuando aparecen términos no lineales, varias variables o ecuaciones acopladas, la dificultad puede aumentar considerablemente.

Por esta razón, en este capítulo no nos centraremos en encontrar soluciones analíticas. Nuestro objetivo será utilizar herramientas **cualitativas y numéricas** para comprender el comportamiento de los modelos. En particular, en esta seccion estudiaremos:

1. **Puntos críticos y estabilidad**, para identificar estados de equilibrio y analizar su comportamiento.
2. **Campos de velocidades**, para visualizar cualitativamente la dinámica de una ecuación diferencial y la naturaleza de los puntos criticos.




### 5.1. Puntos críticos

Consideremos inicialmente una ecuación diferencial autónoma de primer orden,

$$
\frac{dx}{dt}=f(x).
$$

Un **punto crítico** o **punto de equilibrio** es un valor $x^*$ para el cual

$$
f(x^*)=0.
$$

En un punto crítico, la velocidad de cambio de la variable es cero. Por tanto, si el sistema comienza exactamente en $x^*$, permanece en ese estado.

Sin embargo, encontrar los puntos críticos no nos indica por sí solo qué ocurre con las soluciones que comienzan cerca de ellos. Para estudiar su **naturaleza**, podemos analizar el signo de $f(x)$ alrededor de cada punto crítico.

- Si $f(x)>0$, entonces $x$ aumenta.
- Si $f(x)<0$, entonces $x$ disminuye.
- Si las flechas apuntan hacia el punto crítico, este se comporta como un **atractor**.
- Si las flechas apuntan alejándose del punto crítico, este se comporta como un **repulsor**.

Por ejemplo, alrededor de un atractor podemos representar el comportamiento como

```text
        →       →       |       ←       ←
                        x*
                      atractor
```

mientras que alrededor de un repulsor tendríamos

```text
        ←       ←       |       →       →
                        x*
                      repulsor
```

Esta representación permite analizar cualitativamente la dinámica sin resolver la ecuación diferencial.

* #### Ejemplo: modelo logístico

Consideremos el modelo logístico

$$
\frac{dx}{dt}=rx\left(1-\frac{x}{K}\right),
$$

donde $x(t)$ representa una población, $r>0$ es la tasa de crecimiento y $K>0$ es la capacidad de carga del ambiente.

El término

$$
rx
$$

representa el crecimiento natural de la población. Si la población aumenta, también aumenta su crecimiento.

Por otro lado,

$$
1-\frac{x}{K}
$$

representa la limitación producida por los recursos disponibles. Cuando $x$ es pequeña, este factor es cercano a $1$. Cuando $x$ se aproxima a $K$, el factor se aproxima a cero y el crecimiento disminuye.

Los puntos críticos se obtienen imponiendo

$$
rx\left(1-\frac{x}{K}\right)=0.
$$

Por tanto,

$$
x_1^*=0,
\qquad
x_2^*=K.
$$

Para determinar su naturaleza, analizamos el signo de

$$
f(x)=rx\left(1-\frac{x}{K}\right).
$$

Si

$$
0<x<K,
$$

entonces

$$
f(x)>0,
$$

por lo que $x$ aumenta.

Si

$$
x>K,
$$

entonces

$$
f(x)<0,
$$

por lo que $x$ disminuye.

Así, las flechas apuntan hacia $K$ desde ambos lados, mientras que se alejan de $0$ para poblaciones positivas. Por tanto,

- $x^*=0$ es un **repulsor**;
- $x^*=K$ es un **atractor**.

Podemos visualizar este comportamiento mediante la gráfica de $f(x)$.

````python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
r = 0.5
K = 10

# Parámetros de tiempo
t0 = 0
tf = 20
h = 0.01

t = np.arange(t0, tf + h, h)
N = len(t)

# Condiciones iniciales
condiciones_iniciales = [1, 3, 7, 10, 13]

plt.figure(figsize=(9, 5))

for x0 in condiciones_iniciales:

    x = np.zeros(N)
    x[0] = x0

    # Método de Euler
    for n in range(N - 1):
        x[n+1] = x[n] + h*r*x[n]*(1 - x[n]/K)

    plt.plot(t, x, label=fr'$x_0={x0}$')

# Punto crítico x = K
plt.axhline(K, linestyle='--', label=r'$x=K$')

# Punto crítico x = 0
plt.axhline(0, linestyle='--', label=r'$x=0$')

plt.xlabel('$t$')
plt.ylabel('$x(t)$')
plt.title('Soluciones del modelo logístico')
plt.legend()
plt.grid()
plt.show()
````


Ambas gráficas permite identificar los puntos críticos y, observando el signo de $f(x)$ a ambos lados, determinar su naturaleza.



### 5.2. Campo de velocidades de sistemas

Para sistemas de ecuaciones diferenciales, podemos extender esta idea y representar gráficamente la dinámica en el espacio de estados.

Consideremos un sistema de dos ecuaciones,

$$
\begin{aligned}
\frac{dx}{dt} &= f(x,y),\\
\frac{dy}{dt} &= g(x,y).
\end{aligned}
$$

En cada punto $(x,y)$, el sistema determina un vector

$$
\mathbf{V}(x,y)=
\left(
f(x,y),g(x,y)
\right).
$$

Este vector indica la dirección en la que se mueve el sistema cuando se encuentra en ese punto. La representación de estos vectores en una región del plano se denomina **campo de velocidades**. Un punto crítico es precisamente un punto donde la velocidad es cero:

$$
\mathbf{V}(x^*,y^*)=(0,0),
$$

es decir,

$$
f(x^*,y^*)=0,
\qquad
g(x^*,y^*)=0.
$$

El campo de velocidades permite entonces estudiar la **naturaleza de los puntos críticos** observando el comportamiento de las trayectorias cercanas.

Por ejemplo:

- si las trayectorias cercanas se dirigen hacia el punto crítico, este se comporta como un **atractor**;

- si las trayectorias se alejan del punto crítico, se comporta como un **repulsor**;

- si algunas trayectorias se acercan y otras se alejan, podemos observar un comportamiento de tipo **silla**.

Todo este análisis puede realizarse gráficamente, sin necesidad de resolver el sistema ni utilizar autovalores.

> El campo de velocidades permite estudiar la naturaleza de los puntos críticos y visualizar el comportamiento global del sistema.



* #### Ejemplo: Modelo depredaro - presa

Un sistema aparece cuando varias variables evolucionan simultáneamente y sus cambios pueden depender unas de otras. Un ejemplo clásico es el modelo depredador-presa,

$$
\begin{aligned}
\frac{dx}{dt} &= ax-bxy,\\
\frac{dy}{dt} &= -cy+dxy.
\end{aligned}
$$

Aquí,

- $x(t)$ representa la población de presas;
- $y(t)$ representa la población de depredadores.

La primera ecuación contiene el término

$$
ax,
$$

que representa el crecimiento natural de las presas cuando no existen depredadores.

El término

$$
-bxy
$$

representa la disminución de las presas debido a los encuentros con los depredadores. La cantidad de encuentros depende tanto de la población de presas como de la población de depredadores, de ahí la aparición del producto $xy$.

Para los depredadores, el término

$$
-cy
$$

representa su disminución natural cuando no disponen de alimento.

Finalmente,

$$
dxy
$$

representa el crecimiento de la población de depredadores asociado a la disponibilidad de presas.


##### **Puntos críticos**

Los puntos críticos satisfacen simultáneamente

$$
ax-bxy=0,
$$

y

$$
-cy+dxy=0.
$$

Factorizando,

$$
x(a-by)=0,
$$

$$
y(-c+dx)=0.
$$

Por tanto, los puntos críticos son

$$
(x_1^*,y_1^*)=(0,0)
$$

y

$$
(x_2^*,y_2^*)=
\left(
\frac{c}{d},
\frac{a}{b}
\right).
$$

El primer punto representa la ausencia de ambas poblaciones.

El segundo representa un estado en el que las tasas de cambio de ambas poblaciones son simultáneamente cero.


##### **Campo de velocidades**

Podemos construir el campo de velocidades directamente a partir de las dos ecuaciones.

````python
import numpy as np
import matplotlib.pyplot as plt

a = 1.0
b = 0.1
c = 1.5
d = 0.075

x = np.linspace(0, 30, 20)
y = np.linspace(0, 20, 20)

X, Y = np.meshgrid(x, y)

U = a*X - b*X*Y
V = -c*Y + d*X*Y

norma = np.sqrt(U**2 + V**2)

U = U/(norma + 1e-10)
V = V/(norma + 1e-10)

plt.quiver(X, Y, U, V)

plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('Campo de velocidades del modelo depredador-presa')
plt.grid()
plt.show()
````


El campo permite observar cómo cambian las poblaciones dependiendo del estado actual del sistema. Podemos además marcar los puntos críticos y hacer una trazo del campo de velocidades como un flujo usando el graficador `streamplot`:

````python
plt.streamplot(X, Y, U, V) #trazando el campo de velocidades a manera de flujo

plt.scatter(0, 0, color='red', s=60)
plt.scatter(c/d, a/b, color='green', s=60)

plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('Campo de velocidades y puntos críticos')
plt.grid()
plt.show()
````

A partir de la representación gráfica podemos estudiar la naturaleza de los puntos críticos y observar el comportamiento general de las trayectorias.





* #### Ejemplo: campo de velocidades del modelo logístico

Considere el modelo logístico

$$
\frac{dy}{dt}
=
ry\left(1-\frac{y}{K}\right),
$$

donde $y(t)$ representa el tamaño de una población, $r>0$ es la tasa de crecimiento y $K>0$ es la capacidad de carga del ambiente.

Queremos estudiar cualitativamente el comportamiento de las soluciones y determinar hacia dónde evolucionan las poblaciones para diferentes condiciones iniciales.

Para ello, no es necesario resolver la ecuación diferencial. En su lugar, podemos construir su **campo de velocidades**.

El campo de velocidades en el plano $(t,y)$, evaluado en el punto $(t_0,y_0)$, se define mediante el vector normalizado

$$
\left.
\mathbf{u}
\right|_{t_0,y_0}
=
(u_1,u_2)
:=
\frac{1}{\sqrt{V_1^2+V_2^2}}
(V_1,V_2),
$$

donde

$$
V_1=\frac{dt}{dt}=1,
\qquad
V_2=
\left.
\frac{dy}{dt}
\right|_{t_0,y_0}
=
f(t_0,y_0).
$$

Para el modelo logístico,

$$
V_2
=
ry_0\left(1-\frac{y_0}{K}\right).
$$

Así, podemos calcular la dirección de movimiento en cada punto $(t_0,y_0)$ y representar todas esas direcciones en el plano $(t,y)$.

````python
import numpy as np
import matplotlib.pyplot as plt

r = 0.5
K = 100

t = np.linspace(0, 20, 20)
y = np.linspace(0, 150, 20)

T, Y = np.meshgrid(t, y)

V1 = np.ones_like(T)
V2 = r*Y*(1 - Y/K)

norma = np.sqrt(V1**2 + V2**2)

U = V1/norma
V = V2/norma

plt.streamplot(T, Y, U, V)

plt.axhline(K, color='r', linestyle='--', label='$y=K$')

plt.xlabel('$t$')
plt.ylabel('$y$')
plt.title('Campo de velocidades del modelo logístico')
plt.legend()
plt.grid()
plt.show()
````

 El campo de velocidades permite visualizar **todas las posibles trayectorias** que pueden seguir las soluciones, dependiendo de su condición inicial.

Por ejemplo, si

$$
0<y<K,
$$

entonces

$$
\frac{dy}{dt}>0,
$$

por lo que la población crece. Si

$$
y>K,
$$

entonces

$$
\frac{dy}{dt}< 0,
$$

por lo que la población decrece.

En ambos casos, las trayectorias tienden hacia

$$
y=K,
$$

que es un **atractor**.

Por tanto, el campo de velocidades proporciona una visión global del comportamiento del modelo: **no muestra una solución particular, sino las direcciones que permiten construir todas las posibles trayectorias**. Una condición inicial específica seleccionará posteriormente una de ellas.




## 6. Analisis cualitativo de ecuaciones de orden superior

No todos los modelos se describen mediante ecuaciones diferenciales de primer orden. Algunos fenómenos requieren ecuaciones de segundo orden o, en general, de orden superior.

Una ecuación diferencial de orden $n$ puede escribirse como

$$
\frac{d^n x}{dt^n}
=
f\left(
t,x,\frac{dx}{dt},\ldots,\frac{d^{n-1}x}{dt^{n-1}}
\right).
$$

Una ecuación de orden $n$ puede transformarse en un **sistema de $n$ ecuaciones diferenciales de primer orden** mediante la introducción de nuevas variables. Esta transformación permite aplicar directamente los métodos numéricos estudiados para sistemas.

Por ejemplo, una ecuación de segundo orden tiene, p la forma

$$
\frac{d^2x}{dt^2}
=
f\left(t,x,\frac{dx}{dt}\right).
$$

Una ecuación de orden superior puede transformarse en un sistema de ecuaciones de primer orden. Esta transformación es especialmente útil porque los métodos numéricos pueden aplicarse directamente al sistema resultante.



* #### Ejemplo: oscilador armónico

Consideremos el modelo

$$
\frac{d^2x}{dt^2}=-\omega^2x.
$$

Este modelo describe un oscilador armónico ideal, como una masa unida a un resorte.

La variable $x$ representa el desplazamiento respecto a la posición de equilibrio.

El signo negativo es fundamental: indica que la aceleración tiene dirección opuesta al desplazamiento. Por tanto, si la masa se desplaza hacia un lado, aparece una aceleración que tiende a devolverla hacia el equilibrio.



##### **Transformación a un sistema de primer orden:**

Introducimos una nueva variable

$$
v=\frac{dx}{dt}.
$$

Entonces,

$$
\frac{dx}{dt}=v.
$$

Además,

$$
\frac{d^2x}{dt^2}=\frac{dv}{dt}.
$$

Por tanto, la ecuación original se transforma en

$$
\frac{dv}{dt}=-\omega^2x.
$$

Obtenemos así el sistema

$$
\boxed{
\begin{aligned}
\frac{dx}{dt} &= v,\\
\frac{dv}{dt} &= -\omega^2x.
\end{aligned}}
$$

Una ecuación de segundo orden se ha convertido en dos ecuaciones de primer orden.


##### **Punto crítico:**

Para encontrar el punto crítico debemos hacer cero simultáneamente las dos velocidades:

$$
v=0,
$$

y

$$
-\omega^2x=0.
$$

Por tanto,

$$
(x^*,v^*)=(0,0).
$$

Este punto representa la posición de equilibrio con velocidad cero.



##### **Campo de velocidades:**

El campo de velocidades se construye en el plano de estados $(x,v)$. Aqui usaremos la funcion "streamplot" para trazr el campo, la cual es similar a "quiver"


````python
import numpy as np
import matplotlib.pyplot as plt

omega = 1.0

x = np.linspace(-2, 2, 20)
v = np.linspace(-2, 2, 20)

X, V = np.meshgrid(x, v)

U = V
W = -omega**2 * X

norma = np.sqrt(U**2 + W**2)

U = U/(norma + 1e-10)
W = W/(norma + 1e-10)

plt.streamplot(X, V, U, W)

plt.scatter(0, 0, color= 'r', s=60)

plt.xlabel('$x$')
plt.ylabel('$v$')
plt.title('Campo de velocidades del oscilador armónico')
plt.grid()
plt.show()
````




El campo permite interpretar el movimiento sin resolver analíticamente la ecuación.

Por ejemplo:

- cuando $x>0$, la aceleración es negativa;
- cuando $x<0$, la aceleración es positiva;
- cuando $v>0$, la posición aumenta;
- cuando $v<0$, la posición disminuye.

El punto $(0,0)$ es el punto crítico del sistema.

A partir del campo podemos estudiar cualitativamente el comportamiento del oscilador.




## 7. Soluciones numéricas particulares

El campo de velocidades muestra el comportamiento del sistema en una región completa del espacio de estados.

Sin embargo, podemos estar interesados en una pregunta diferente:

> ¿Qué trayectoria sigue el sistema si comienza en un estado inicial determinado?

Por ejemplo, supongamos que

$$
x(0)=20,
\qquad
y(0)=10.
$$

La condición inicial determina un punto específico del espacio de estados.

A partir de ese punto, el campo de velocidades indica hacia dónde comienza a moverse el sistema. Al seguir sucesivamente estas direcciones obtenemos una **trayectoria particular**.

Los métodos numéricos permiten aproximar esa trayectoria.

Esta diferencia es fundamental:

$$
\boxed{
\text{Campo de velocidades}
\quad\longrightarrow\quad
\text{comportamiento global}
}
$$

mientras que

$$
\boxed{
\text{Método numérico + condición inicial}
\quad\longrightarrow\quad
\text{una trayectoria particular}
}
$$

A continuacion discutiremos dos importantes metodos para encontrar soluciones numericas de ecuaciones o sistemas de ecuaciones diferenciales ordinarias.



### 7.1. Método de Euler

Una vez comprendido el comportamiento cualitativo de un modelo, podemos utilizar métodos numéricos para calcular trayectorias particulares.

Consideremos nuevamente una ecuación de primer orden,

$$
\frac{dx}{dt}=f(t,x),
\qquad
x(t_0)=x_0.
$$

Dividimos el intervalo de tiempo en pasos de tamaño $h$:

$$
t_{n+1}=t_n+h.
$$

El método de Euler parte de la aproximación

$$
\frac{x_{n+1}-x_n}{h}
\approx
f(t_n,x_n).
$$

Despejando,

$$
\boxed{
x_{n+1}
=
x_n+h f(t_n,x_n)
}.
$$

Geométricamente, Euler utiliza la pendiente en el punto actual para avanzar hasta el siguiente punto.



##### **Deducción a partir de Taylor:**

La expansión de Taylor alrededor de $t_n$ es

$$
x(t_n+h)
=
x(t_n)
+h x'(t_n)
+\frac{h^2}{2}x''(t_n)
+\mathcal{O}(h^3).
$$

Como

$$
x'(t_n)=f(t_n,x_n),
$$

tenemos

$$
x(t_n+h)
=
x(t_n)
+h f(t_n,x_n)
+\frac{h^2}{2}x''(t_n)
+\mathcal{O}(h^3).
$$

El método de Euler conserva solamente los dos primeros términos:

$$
x_{n+1}
=
x_n+h f(t_n,x_n).
$$

Por tanto, el primer término que se omite es proporcional a $h^2$. El error local es entonces

$$
\mathcal{O}(h^2),
$$

mientras que el error global es

$$
\mathcal{O}(h).
$$

Por esta razón, Euler es un método de **orden 1**.



### 7.2 Método de Adams-Bashforth de dos pasos

Podemos mejorar la aproximación de Euler utilizando información del paso anterior. Partimos de la expansión de Taylor alrededor de $t_n$:

$$
x_{n+1}
=
x_n
+
h x'(t_n)
+
\frac{h^2}{2}x''(t_n)
+
\mathcal{O}(h^3).
$$

Como

$$
x'(t_n)=f_n,
$$

tenemos

$$
x_{n+1}
=
x_n
+
h f_n
+
\frac{h^2}{2}x''(t_n)
+
\mathcal{O}(h^3).
$$

Para aproximar la segunda derivada utilizamos una diferencia finita hacia atrás:

$$
x''(t_n)
\approx
\frac{x'(t_n)-x'(t_{n-1})}{h}.
$$

Como

$$
x'(t_n)=f_n,
\qquad
x'(t_{n-1})=f_{n-1},
$$

obtenemos

$$
x''(t_n)
\approx
\frac{f_n-f_{n-1}}{h}.
$$

Sustituyendo en la expansión de Taylor:

$$
x_{n+1}
=
x_n
+
h f_n
+
\frac{h}{2}(f_n-f_{n-1}),
$$

por lo que

$$
\boxed{
x_{n+1}
=
x_n+
\frac{h}{2}
\left(3f_n-f_{n-1}\right)
}.
$$

Esta es la fórmula del **método de Adams-Bashforth de dos pasos**.



##### **¿Por qué es de orden 2?:**

A diferencia de Euler, que utiliza únicamente el término de primer orden,

$$
x_{n+1}=x_n+h f_n,
$$

Adams-Bashforth incorpora una aproximación del término de segundo orden mediante la información del paso anterior. Por ello, su error local es

$$
\mathcal{O}(h^3),
$$

mientras que su error global es

$$
\boxed{\mathcal{O}(h^2)}.
$$

Por tanto, Adams-Bashforth de dos pasos es un método de **orden 2**.

| Método | Información utilizada | Error global | Orden |
|---|---|---|---:|
| Euler | $f_n$ | $\mathcal{O}(h)$ | 1 |
| Adams-Bashforth de dos pasos | $f_n,\ f_{n-1}$ | $\mathcal{O}(h^2)$ | 2 |

Adams-Bashforth necesita dos valores iniciales, $x_0$ y $x_1$. El valor $x_0$ proviene de la condición inicial, mientras que $x_1$ puede obtenerse mediante un paso de Euler. Por esta razón, aunque Euler no sea especialmente eficiente cuando se requiere alta precisión, es importante como método inicial para poner en marcha Adams-Bashforth.




### 7.3 Ejemplo: Comparación entre Euler y Adams-Bashforth

Consideremos el modelo

$$
\frac{dx}{dt}=rx,
$$

con

$$
r=0.2,
\qquad
x(0)=10.
$$

Para Euler,

$$
x_{n+1}
=
x_n+hrx_n.
$$

Para Adams-Bashforth de dos pasos,

$$
x_{n+1}
=
x_n+
\frac{h}{2}
\left(
3rx_n-rx_{n-1}
\right).
$$

Adams-Bashforth necesita dos valores iniciales. Por ello, utilizaremos Euler para obtener el segundo valor y luego continuaremos con Adams-Bashforth.

````python
import numpy as np
import matplotlib.pyplot as plt

r = 0.2
h = 0.2
N = 50

t = np.arange(N + 1)*h

euler = np.zeros(N + 1)
adams = np.zeros(N + 1)

euler[0] = 10
adams[0] = 10

# Primer paso de Adams-Bashforth mediante Euler
adams[1] = adams[0] + h*r*adams[0]

# Método de Euler
for n in range(N):
    euler[n + 1] = euler[n] + h*r*euler[n]

# Adams-Bashforth de dos pasos
for n in range(1, N):
    fn = r*adams[n]
    fn_1 = r*adams[n - 1]

    adams[n + 1] = adams[n] + h/2*(3*fn - fn_1)

plt.plot(t, euler, label='Euler')
plt.plot(t, adams, label='Adams-Bashforth 2')

plt.xlabel('$t$')
plt.ylabel('$x(t)$')
plt.title('Comparación entre Euler y Adams-Bashforth')
plt.legend()
plt.grid()
plt.show()
````

La comparación permite observar que, para un mismo tamaño de paso, el método de Adams-Bashforth de dos pasos puede producir una aproximación más precisa que Euler.

En general,

$$
\text{Euler:}\qquad
\text{error global}=\mathcal{O}(h),
$$

mientras que

$$
\text{Adams-Bashforth 2:}\qquad
\text{error global}=\mathcal{O}(h^2).
$$

Por tanto, al disminuir el tamaño del paso, el error de Adams-Bashforth disminuye más rápidamente.



**Nota importante: **
Aunque el método de Euler es sencillo y, en general, no resulta suficientemente preciso para resolver problemas que requieren mayor exactitud, su estudio es fundamental porque constituye la base para comprender métodos de orden superior. En particular, el método de Adams-Bashforth de dos pasos necesita dos valores iniciales para comenzar el proceso, y uno de ellos puede obtenerse mediante un primer paso de Euler. Por esta razón, aunque Euler tenga limitaciones prácticas, es un método necesario como punto de partida para introducir y aplicar métodos más precisos.



### 7.4. Ejemplo: Resolviendo un sistema de ecuaciones

Usando la segunda ley de Newton se obtiene la ecuación de movimiento del péndulo simple:

$$
\frac{d^2\theta}{dt^2}
+
\frac{g}{l}\sin(\theta)
=0.
$$

Esta ecuación es **no lineal** debido al término $\sin(\theta)$ y, en general, no es necesario buscar una solución analítica para estudiar su comportamiento. Podemos resolverla numéricamente.

Como es una ecuación diferencial de segundo orden, introducimos la variable

$$
\omega=\frac{d\theta}{dt}.
$$

De esta forma obtenemos el sistema de primer orden

$$
\boxed{
\begin{aligned}
\frac{d\theta}{dt} &= \omega,\\
\frac{d\omega}{dt} &= -\frac{g}{l}\sin(\theta).
\end{aligned}}
$$

Consideremos las condiciones iniciales

$$
\theta(0)=\frac{\pi}{4},
\qquad
\omega(0)=0.
$$

Tomemos

$$
g=9.81\ \text{m/s}^2,
\qquad
l=1\ \text{m}.
$$



##### **Método de Euler:**

Para un sistema

$$
\frac{dx}{dt}=f(x,y),
\qquad
\frac{dy}{dt}=g(x,y),
$$

el método de Euler está dado por

$$
x_{n+1}=x_n+h f(x_n,y_n),
$$

$$
y_{n+1}=y_n+h g(x_n,y_n).
$$

En nuestro caso,

$$
f(\theta,\omega)=\omega,
\qquad
g(\theta,\omega)=-\frac{g}{l}\sin(\theta).
$$

Por tanto,

$$
\boxed{
\begin{aligned}
\theta_{n+1}
&=
\theta_n+h\omega_n,\\
\omega_{n+1}
&=
\omega_n
-h\frac{g}{l}\sin(\theta_n).
\end{aligned}}
$$

##### **Método de Adams-Bashforth de dos pasos:**

Para el mismo sistema, Adams-Bashforth utiliza los valores de la función en los dos últimos pasos:

$$
\theta_{n+1}
=
\theta_n+
\frac{h}{2}
\left(
3\omega_n-\omega_{n-1}
\right),
$$

y

$$
\omega_{n+1}
=
\omega_n+
\frac{h}{2}
\left[
-3\frac{g}{l}\sin(\theta_n)
+
\frac{g}{l}\sin(\theta_{n-1})
\right].
$$

Como Adams-Bashforth necesita dos valores iniciales, calculamos el primer paso utilizando Euler.


````python

### Implementación en Python

import numpy as np
import matplotlib.pyplot as plt

# Parámetros
g = 9.81
l = 1.0

# Paso y tiempo
h = 0.02
T = 10
N = int(T/h)

t = np.arange(N + 1)*h

# Condiciones iniciales
theta0 = np.pi/4
omega0 = 0

# Arreglos
theta_euler = np.zeros(N + 1)
omega_euler = np.zeros(N + 1)

theta_adams = np.zeros(N + 1)
omega_adams = np.zeros(N + 1)

theta_euler[0] = theta0
omega_euler[0] = omega0

theta_adams[0] = theta0
omega_adams[0] = omega0

# Primer paso de Adams-Bashforth usando Euler
theta_adams[1] = theta_adams[0] + h*omega_adams[0]

omega_adams[1] = (
    omega_adams[0]
    - h*(g/l)*np.sin(theta_adams[0])
)

# Método de Euler
for n in range(N):

    theta_euler[n+1] = (
        theta_euler[n]
        + h*omega_euler[n]
    )

    omega_euler[n+1] = (
        omega_euler[n]
        - h*(g/l)*np.sin(theta_euler[n])
    )

# Adams-Bashforth de dos pasos
for n in range(1, N):

    theta_adams[n+1] = (
        theta_adams[n]
        + h/2*(3*omega_adams[n]
               - omega_adams[n-1])
    )

    omega_adams[n+1] = (
        omega_adams[n]
        + h/2*(
            -3*(g/l)*np.sin(theta_adams[n])
            +(g/l)*np.sin(theta_adams[n-1])
        )
    )

# Gráfica
plt.plot(t, theta_euler, label='Euler')
plt.plot(t, theta_adams, label='Adams-Bashforth 2')

plt.xlabel('$t$')
plt.ylabel(r'$\theta(t)$')
plt.title('Péndulo simple: comparación de métodos numéricos')
plt.legend()
plt.grid()
plt.show()
````

Desde el punto de vista físico, el péndulo ideal descrito por las leyes de Newton debe conservar su energía mecánica. Por tanto, en ausencia de fuerzas externas y de fricción, la amplitud de sus oscilaciones debe mantenerse aproximadamente constante: el péndulo no puede generar energía por sí mismo ni amplificar progresivamente sus oscilaciones. Una solución numérica que muestre un crecimiento artificial de la amplitud no representa correctamente el comportamiento físico del sistema, sino que refleja la acumulación de error numérico del método empleado.

En este ejemplo, el método de Euler puede producir precisamente este comportamiento no físico, haciendo que la amplitud de las oscilaciones aumente progresivamente. En cambio, el método de Adams-Bashforth de dos pasos proporciona una aproximación mucho más adecuada, manteniendo las oscilaciones con una amplitud mucho más estable. Esto muestra una de las limitaciones prácticas de Euler y la ventaja de utilizar un método de mayor orden como Adams-Bashforth para obtener soluciones numéricas que representen mejor la dinámica del sistema.



### 7.5. Ejemplo: Trayectorias particulares en el campo de velocidades

Para complementar el análisis, construiremos ahora el campo de velocidades en el plano de fase $(\theta,\omega)$ y sobrepondremos las trayectorias obtenidas mediante los métodos de Euler y Adams-Bashforth. De esta manera, podremos visualizar el comportamiento global del sistema y, al mismo tiempo, comparar cómo cada método numérico aproxima la trayectoria correspondiente a la condición inicial elegida. Esto permitirá identificar gráficamente las diferencias entre ambos métodos y apreciar cuál representa mejor la dinámica física del péndulo.

````python
# Campo de velocidades y trayectorias numéricas

import numpy as np
import matplotlib.pyplot as plt

# Parámetros
g = 9.81
l = 1.0

# Malla en el plano de fase
theta = np.linspace(- np.pi,  np.pi, 25)
omega = np.linspace(-2*np.pi, 2*np.pi, 25)

Theta, Omega = np.meshgrid(theta, omega)

# Campo de velocidades
U = Omega
V = -(g/l)*np.sin(Theta)

# Normalización para visualizar mejor las flechas
norma = np.sqrt(U**2 + V**2)

U = U/(norma + 1e-10)
V = V/(norma + 1e-10)

# Gráfica
plt.figure(figsize=(10, 6))

plt.streamplot(
    Theta, Omega,
    U, V)

# Trayectoria obtenida con Euler
plt.plot(
    theta_euler,
    omega_euler,
    label='Euler'
)

# Trayectoria obtenida con Adams-Bashforth
plt.plot(
    theta_adams,
    omega_adams,
    label='Adams-Bashforth 2'
)

# Condición inicial
plt.scatter(
    theta0,
    omega0, color='red',
    s=60,
    label='Condición inicial'
)

plt.xlabel(r'$\theta$')
plt.ylabel(r'$\omega$')
plt.title('Campo de velocidades y trayectorias del péndulo')
plt.legend()
plt.grid()
plt.show()
````

El campo de velocidades representa las posibles trayectorias del sistema para diferentes condiciones iniciales, mientras que cada curva superpuesta corresponde a una condición inicial particular. En este caso, podemos comparar directamente cómo Euler y Adams-Bashforth aproximan esa misma trayectoria en el espacio de fase $(\theta,\omega)$. El comportamiento esperado del péndulo ideal corresponde a trayectorias cerradas alrededor del punto de equilibrio $(0,0)$; por ello, las desviaciones que produzca Euler pueden interpretarse como errores numéricos y no como un comportamiento físico del péndulo.

## 9.  Ideas principales de este capitulo
En este capítulo hemos pasado de modelos discretos

$$
x_{n+1}=f(x_n)
$$

a modelos continuos de la forma

$$
\frac{dx}{dt}=f(t,x).
$$

Hemos visto que las ecuaciones diferenciales permiten representar tasas de cambio y que los problemas de valor inicial incorporan las condiciones particulares del sistema. También hemos estudiado modelos de crecimiento y decaimiento, regulación poblacional, interacción entre especies y sistemas de ecuaciones diferenciales.

Una idea especialmente importante es que existen diferentes niveles de análisis. Podemos comenzar con una ecuación diferencial:

$$
\frac{dx}{dt}=f(x),
$$

y estudiar sus puntos de equilibrio y el signo de $f(x)$ sin encontrar explícitamente una solución. También podemos buscar una solución analítica mediante técnicas como la separación de variables.

Finalmente, cuando una solución explícita no está disponible o resulta difícil de obtener, podemos utilizar métodos numéricos:

$$
\boxed{
\text{modelo}
\rightarrow
\text{ecuación diferencial}
\rightarrow
\text{análisis cualitativo}
\rightarrow
\text{solución analítica o numérica}
}
$$

En modelos con muchas variables, esta estructura conduce naturalmente al estudio de sistemas dinámicos, análisis de estabilidad, métodos numéricos para sistemas de ecuaciones y, posteriormente, a modelos más sofisticados como ecuaciones diferenciales parciales.

Los modelos continuos constituyen así un punto de encuentro entre el cálculo, las ecuaciones diferenciales, el análisis cualitativo, la computación científica y el modelamiento matemático.
